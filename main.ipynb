# **1. Introduction**

The US cell phone service market is incredibly saturated and most adults already have a provider. Because of this, telecom companies cannot rely on steady new customer growth and instead depend heavily on customer retention. Losing a customer means losing all future monthly revenue and often paying additional costs to acquire a replacement. That makes the ability of being able to identify customers who are likely to leave strategically important for profits, scale, and efficency.

This project analyzes the IBM Telco Customer Churn dataset which simulates a telecom company’s customer base. Each row describes one customer and includes demographic information, services used, billing details, and whether the customer left in the past month. The goal of this work is to identify patterns associated with customer churn and to evaluate two predictive models (logistic regression and a pruned decision tree) to see how well they predict high-risk customers. The analysis is exploratory and aims to highlight insights that could guide real-world retention strategies.

# **2. Background and Motivation**

Telecom companies face intense competition and limited opportunities for expansion through new customer acquisition. Retaining existing customers is generally more cost-effective than recruiting new ones. Predictive churn modeling provides a way to flag high-risk customers early, allowing companies to offer targeted support or incentives.

This project focuses on three main questions:

1. Are newer customers (shorter tenure) more likely to churn?
2. How do pricing and contract structure relate to churn risk?
3. How well do two simple models of logistic regression and a pruned decision tree distinguish churners from non-churners?

The dataset includes demographic attributes, subscribed services, contract type, billing information, and whether the customer churned in the past month. Approximately three-quarters of customers do not churn, producing a clear class imbalance that affects model evaluation. Because accuracy can be misleading in this context, we rely on AUC (Area Under the ROC Curve) as a complementary metric.

These variables allow us to explore how customer characteristics relate to churn and whether simple supervised learning models can predict churn reliably enough to assist with retention strategies.


Churn is the rate at which customers leave a company over a given time period.
Churn rate = (Number of customers who leave during a period) ÷ (Total customers at the start of the period)
A company has high churn risk if they are likely to leave the company soon. A customer is low churn risk if they are unlikely to leave the company soon
In this dataset, churn is recorded as either “Yes” for customers who left in the past month or “No” for those who stayed. About three-quarters of the customers do not churn, which creates an imbalance that makes prediction more difficult. Any model must do more than simply predict “No” for every customer.

# **4. Modeling Approach**
   
Two predictive models were used in this analysis: logistic regression and a decision tree classifier. These models are transparent, easy to interpret, and consistent with the exploratory goals of the project.

### **Logistic Regression**
Trained on all columns except customerID and Churn. We use this model to predict churn. Logistic regression provides a simple baseline and outputs estimated probabilities of churn. All predictor variables were included except customerID and the churn label. Categorical variables were converted into indicator variables. Logistic regression performs well on many churn datasets and offers insight into how each variable affects churn risk. We split the data into a training set and a 20 percent held-out test set using stratified sampling to preserve the churn proportion. Logistic regression was fit on the training data, with all categorical features encoded as indicator variables. We evaluated the model using test accuracy and the area under the ROC curve (AUC), because accuracy alone can be misleading with imbalanced outcomes.

### **Decision Tree**
Trained on all columns except customer ID and Churn. I decided to leave all hyperparameters at their default and only tune ccp alpha.  I perform 5-fold Cross Validation using the accuracy scoring metric. I explored and visualized the ccp_alpha space by plugging in different ranges for ccp_alpha. I also experimented with different levels of ccp_alpha coarseness by varying the spacing between adjacent ccp_alpha values. All of this was done iteratively using np.linspace. After looking at the visualizations, I noticed a common shape in the plot: an increase in CV accuracy, followed by a flatline, followed by a decrease. I then zoomed in on the region where CV accuracy appeared to be largest. After looking at several plots, I eventually settled with a range of ccp_alpha from 0 to 0.035 and a coarseness of 100 evenly spaced values, as I determined it would be sufficient to capture the optimal ccp_alpha. The optimal ccp_alpha ended up being 0.0032. The accuracy of the CART model on the test set is 0.7818. Decision trees offer a more flexible approach. They can handle nonlinear interactions without manually constructing interaction terms. However, decision trees can overfit, so the model was pruned using cost complexity pruning. A grid of 100 values for the pruning parameter, ccp_alpha, ranging from 0 to 0.035, was evaluated using 5-fold cross validation. The pattern of results was clear: accuracy rose quickly from zero, stayed flat for a wide range of ccp_alpha values, and then declined when pruning became too aggressive. The best value was ccp_alpha = 0.0032. For the decision tree we used cost-complexity pruning. We performed 5-fold cross-validation over a grid of candidate ccp_alpha values between 0 and 0.035, using accuracy as the scoring metric. Cross-validated accuracy increased from very small ccp_alpha values, stayed flat over a broad middle region, and then declined once the tree was over-pruned. The best value, ccp_alpha = 0.0032, defines the pruned tree we evaluate on the test set.

The AUC (Area Under the Curve) of both models, which measures how well the two models discriminate, were computed. The logistic regression model outperformed the tree model on both AUC and accuracy. Accuracy measures the overall fraction of correctly classified customers, while AUC measures how well the model ranks churners above non-churners across all probability thresholds. With an imbalanced outcome like churn, AUC is an important complement to accuracy.

# **5. Results**
   
Logistic regression achieved a test accuracy of about 0.7967 and an AUC of about 0.8305. The confusion matrix shows that logistic regression correctly identifies many non-churners while also catching a meaningful portion of churners (see figure 3), although some churners are still missed.
The decision tree achieved a test accuracy of about 0.7818 and an AUC of about 0.7937. It correctly classifies many non-churners and produces fewer false positives than logistic regression, but it misses more true churners. This makes the tree more conservative. While its accuracy is only slightly lower than that of logistic regression, its lower AUC and higher number of missed churners reduce its usefulness for retention strategies.
Overall, logistic regression performs better on this dataset. It is more balanced across the two classes and identifies more churners, which is often the most important outcome for a business seeking to prevent customer loss.

# **6. Interpretation and Implications**

Insights from EDA and predictive modeling:

- Customers with short tenure are significantly more likely to churn.  
- Customers with higher monthly charges show increased churn rates, suggesting dissatisfaction with pricing.  
- Contract type is one of the strongest predictors: month-to-month customers churn far more often than those on longer contracts.  
- Service bundles such as streaming, security, and technical support may increase customer stickiness.

These findings suggest the following strategies for retention:

1. Prioritize engagement for customers in their first year.  
2. Offer targeted rate adjustments or discounts to high-bill customers.  
3. Encourage customers to switch to one-year or two-year contracts.  
4. Design bundled service packages that meet customer needs and reduce churn.

# **7. Assumptions and Limitations**

### **7.1 Assumptions**
- Each row corresponds to a unique customer account.  
- Churn labels accurately reflect customer departures.  
- Tenure, charges, contract type, and service features are measured reliably.  
- Dropping rows with missing TotalCharges does not introduce systematic bias.  
- The snapshot is representative of customer behavior during the period analyzed.

### **7.2 Limitations**
- This is observational data (no causal claims can be made).  
- The dataset is imbalanced, making accuracy insufficient without AUC.  
- Real telecom markets shift rapidly, so models must be re-trained regularly.  
- Only two simple models were used in this project so, more complex models could improve prediction % but reduce interpretability, reproducibility, and increase complexity.
