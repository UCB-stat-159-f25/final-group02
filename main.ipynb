**1. Introduction**

   
The US cell phone service market is incredibly saturated. Almost every adult already has a provider, so telecom companies cannot rely on constant new customer growth. Instead, they focus on retaining the customers they already have. Losing a customer means losing all future monthly revenue and often paying more again in marketing costs to replace them. Predicting which customers are likely to leave is therefore important for the industry.
This project examines the IBM Telco Customer Churn dataset, which simulates a telecom company’s customer base. Each row describes a single customer, including demographic information, the services they subscribe to, their billing details, and whether they left in the past month. Based on the dataset we want to answer questions such as “What customer characteristics strongly predict churn?” The goal of the project is to identify patterns related to churn and to evaluate two predictive models, logistic regression and a decision tree, to see how well they detect high-risk customers. The analysis is exploratory and focuses on understanding patterns that could guide real-world retention strategies.


**2. Background and Motivation**


Telecom companies operate in a competitive environment with little room for growth through new customer acquisition. Retaining existing customers is often more cost-effective than attracting new ones. Having a way to flag customers who are likely to leave allows companies to send targeted offers, improve customer support, or adjust plans to keep them from switching providers.
In this project we focus on three main questions:
Are newer customers (shorter tenure) more likely to churn than long-tenure customers?


How are pricing and contract structure (monthly charges and contract type) related to churn risk?


How well can simple predictive models (logistic regression and a pruned decision tree) distinguish between customers who churn and those who stay?
Churn is the rate at which customers leave a company over a given time period.
Churn rate = (Number of customers who leave during a period) ÷ (Total customers at the start of the period)
A company has high churn risk if they are likely to leave the company soon. A customer is low churn risk if they are unlikely to leave the company soon
In this dataset, churn is recorded as either “Yes” for customers who left in the past month or “No” for those who stayed. About three-quarters of the customers do not churn, which creates an imbalance that makes prediction more difficult. Any model must do more than simply predict “No” for every customer.

**3. Exploratory Data Analysis and Key Observations**

Before fitting models, I examined the distributions and relationships among the variables.
The strongest numerical relationship in the correlation matrix is between tenure and TotalCharges. This is expected because total charges accumulate over the time a customer stays. MonthlyCharges correlate moderately with TotalCharges but only weakly with tenure, which means that customers pay very different monthly amounts regardless of how long they have been with the company.
Tenure has a distinctive shape in its histogram: a large group of new customers with very low tenure and another group of long-time customers with tenure near the maximum of 72 months. MonthlyCharges show a wide spread from low to high bill amounts, reflecting a variety of plans and service bundles.
Churn patterns also emerge clearly in the exploratory plots. Customers who churn tend to have lower tenure, while those who stay have longer tenures on average. Monthly charges are usually higher for churned customers, suggesting that higher bills may lead to dissatisfaction. Contract type is one of the strongest indicators of churn. Month-to-month customers churn at far higher rates than customers on one-year or two-year contracts.
These patterns support the idea that contract structure, pricing, and early customer experience play a significant role in churn. These findings also helped justify including all features in the predictive models.

**4. Modeling Approach**
   
Two predictive models were used in this analysis: logistic regression and a decision tree classifier. These models are transparent, easy to interpret, and consistent with the exploratory goals of the project.

A logistic regression model trained on all columns except customerID and Churn. We use this model to predict churn. Logistic regression provides a simple baseline and outputs estimated probabilities of churn. All predictor variables were included except customerID and the churn label. Categorical variables were converted into indicator variables. Logistic regression performs well on many churn datasets and offers insight into how each variable affects churn risk. We split the data into a training set and a 20 percent held-out test set using stratified sampling to preserve the churn proportion. Logistic regression was fit on the training data, with all categorical features encoded as indicator variables. We evaluated the model using test accuracy and the area under the ROC curve (AUC), because accuracy alone can be misleading with imbalanced outcomes.

A decision tree algorithm trained on all columns except customer ID and Churn. I decided to leave all hyperparameters at their default and only tune ccp alpha.  I perform 5-fold Cross Validation using the accuracy scoring metric. I explored and visualized the ccp_alpha space by plugging in different ranges for ccp_alpha. I also experimented with different levels of ccp_alpha coarseness by varying the spacing between adjacent ccp_alpha values. All of this was done iteratively using np.linspace. After looking at the visualizations, I noticed a common shape in the plot: an increase in CV accuracy, followed by a flatline, followed by a decrease. I then zoomed in on the region where CV accuracy appeared to be largest. After looking at several plots, I eventually settled with a range of ccp_alpha from 0 to 0.035 and a coarseness of 100 evenly spaced values, as I determined it would be sufficient to capture the optimal ccp_alpha. The optimal ccp_alpha ended up being 0.0032. The accuracy of the CART model on the test set is 0.7818. Decision trees offer a more flexible approach. They can handle nonlinear interactions without manually constructing interaction terms. However, decision trees can overfit, so the model was pruned using cost complexity pruning. A grid of 100 values for the pruning parameter, ccp_alpha, ranging from 0 to 0.035, was evaluated using 5-fold cross validation. The pattern of results was clear: accuracy rose quickly from zero, stayed flat for a wide range of ccp_alpha values, and then declined when pruning became too aggressive. The best value was ccp_alpha = 0.0032. For the decision tree we used cost-complexity pruning. We performed 5-fold cross-validation over a grid of candidate ccp_alpha values between 0 and 0.035, using accuracy as the scoring metric. Cross-validated accuracy increased from very small ccp_alpha values, stayed flat over a broad middle region, and then declined once the tree was over-pruned. The best value, ccp_alpha = 0.0032, defines the pruned tree we evaluate on the test set.

The AUC (Area Under the Curve) of both models, which measures how well the two models discriminate, were computed. The logistic regression model outperformed the tree model on both AUC and accuracy. Accuracy measures the overall fraction of correctly classified customers, while AUC measures how well the model ranks churners above non-churners across all probability thresholds. With an imbalanced outcome like churn, AUC is an important complement to accuracy.


**5. Results**
   
Logistic regression achieved a test accuracy of about 0.7967 and an AUC of about 0.8305. The confusion matrix shows that logistic regression correctly identifies many non-churners while also catching a meaningful portion of churners (see figure 3), although some churners are still missed.
The decision tree achieved a test accuracy of about 0.7818 and an AUC of about 0.7937. It correctly classifies many non-churners and produces fewer false positives than logistic regression, but it misses more true churners. This makes the tree more conservative. While its accuracy is only slightly lower than that of logistic regression, its lower AUC and higher number of missed churners reduce its usefulness for retention strategies.
Overall, logistic regression performs better on this dataset. It is more balanced across the two classes and identifies more churners, which is often the most important outcome for a business seeking to prevent customer loss.


**6. Interpretation and Implications**
   
The combined findings from exploratory analysis and predictive modeling reveal several important points:
Short tenure is a major signal for churn. New customers are far more likely to leave.
 High monthly charges are associated with higher churn. Pricing may play a role in dissatisfaction.
 Long contracts predict lower churn. Contract structure is one of the most effective tools for retention.
 Service bundles such as technical support or streaming options may increase customer stickiness.
 
For a telecom company, these findings suggest several possible retention strategies:
Focus support and incentives on new customers within their first year.
 Consider offering targeted discounts or rate adjustments for high-bill customers.
 Promote one-year and two-year contracts to reduce churn risk.
 Encourage customers to adopt more services or bundles that fit their needs.
 
Even moderate predictive performance can be useful. A model that identifies a portion of high-risk customers can guide limited retention resources more effectively than random outreach.


**7. Assumptions and Limitations**
   
***7.1 Assumptions about the data***

For this analysis we treat the IBM Telco dataset as a cross-sectional snapshot of one month of customer behavior. We assume:
• Each row represents one independent customer account.
• The churn label is correctly recorded, so “Yes” truly means the customer left during the reference month.
• The covariates (tenure, charges, contract type, and service indicators) were measured without major error.
• Dropping the small number of rows with missing TotalCharges does not systematically bias the sample, and the remaining 7032 customers are representative of the overall customer base.
• The relationships between the predictors and churn are approximately stable over time, so patterns learned from this snapshot are informative for near-future customers.
Under these assumptions, it is reasonable to treat the dataset as a basis for exploratory modeling and to evaluate classification performance with standard train/test splits and cross-validation.

***7.2 Limitations***

This project is based on observational data. Customers were not randomly assigned to contract types or pricing plans. Many unmeasured factors, such as regional service quality or customer support interactions, could influence churn. Because of this, the relationships we observe are descriptive and not causal.
Another limitation is the imbalance in the dataset. About three-quarters of customers do not churn, so accuracy alone can be misleading. This is why AUC and confusion matrices were also used to evaluate performance.
We assume the churn labels are accurate and that the dataset reflects stable customer behavior. Real companies face evolving market conditions, new competitors, and changing pricing structures. Models would need regular updates in such settings.
The analysis focuses on two simple models. More advanced models could improve prediction, but the goal here was transparency and reproducibility rather than maximum performance.

**9. Conclusions**

This project shows that simple predictive models can extract meaningful structure from the Telco churn dataset. Logistic regression performs slightly better than the decision tree in both accuracy and the ability to distinguish churners from non-churners. The exploratory analysis highlights the importance of tenure, contract type, and monthly charges. These patterns align with common business practices in the telecom industry.
Even though the dataset is simulated and the analysis is exploratory, the results offer insight into effective retention strategies. In a saturated market, being able to identify high-risk customers can provide a significant advantage.
